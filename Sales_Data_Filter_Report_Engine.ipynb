# **Sales Data Filter & Report Engine**

Objective: Build a NumPy-based sales analysis system that simulates annual sales for 10 stores across 12 months and generates useful reports through array operations, slicing, boolean masking and fancy indexing.
Technologies: Python, NumPy, Google Colab

Key Features:
- Annual sales simulation
- Quarterly sales analysis
- Monthly and store-wise reports
- Sales target filtering
- Best-performing store and month analysis
- Growth target generation



```python
import numpy as np

np.random.seed(42)

months = np.arange(1, 13)
stores = np.arange(1, 11)

# Sales data: 10 stores x 12 months
sales = np.random.randint(10000, 50001, size=(10, 12))

print("SALES DATA")
print(sales)

print("\nShape:", sales.shape)
print("Dimensions:", sales.ndim)
print("Total values:", sales.size)
print("Data type:", sales.dtype)
```

```python
monthly_sales = np.sum(sales, axis=0)

print("MONTHLY SALES REPORT")
for i in range(12):
    print(f"Month {months[i]}: ₹{monthly_sales[i]:,}")

print("\nTotal Annual Sales: ₹", np.sum(sales))
print("Average Monthly Sales: ₹", np.mean(monthly_sales))
```

```python
store_totals = np.sum(sales, axis=1)
store_averages = np.mean(sales, axis=1)

print("STORE-WISE SALES REPORT")

for i in range(10):
    print(
        f"Store {stores[i]}: "
        f"Total = ₹{store_totals[i]:,}, "
        f"Monthly Average = ₹{store_averages[i]:,.2f}"
    )

best_store = np.argmax(store_totals)
worst_store = np.argmin(store_totals)

print("\nBest Performing Store:", stores[best_store])
print("Lowest Performing Store:", stores[worst_store])
```

```python
q1 = sales[:, 0:3]
q2 = sales[:, 3:6]
q3 = sales[:, 6:9]
q4 = sales[:, 9:12]

print("QUARTERLY SALES REPORT")
print("Q1 Total:", np.sum(q1))
print("Q2 Total:", np.sum(q2))
print("Q3 Total:", np.sum(q3))
print("Q4 Total:", np.sum(q4))

# Sales of Store 1 across all months
print("\nStore 1 Sales:", sales[0, :])

# First three months of Store 1
print("Store 1 Q1 Sales:", sales[0, 0:3])
```

```python
target = 300000

# Months exceeding the target
high_sales_months = monthly_sales[monthly_sales > target]

print("Months with sales above ₹", target)
print(high_sales_months)

# Stores below average annual sales
average_store_sales = np.mean(store_totals)

below_average = store_totals[store_totals < average_store_sales]

print("\nAverage Annual Sales per Store:", average_store_sales)
print("Sales of Below-Average Stores:", below_average)

# Combined conditions: months above ₹250,000 AND below ₹450,000
mask = (monthly_sales > 250000) & (monthly_sales < 450000)

print("\nMonths within the selected sales range:")
print(months[mask])

# OR condition
high_or_low = (monthly_sales > 450000) | (monthly_sales < 250000)
print("\nMonths outside the selected range:")
print(months[high_or_low])

# NOT condition
not_high = ~(monthly_sales > 400000)
print("\nMonths not exceeding ₹400,000:")
print(months[not_high])
```

```python
selected_indices = [0, 3, 6]

selected_stores = sales[selected_indices, :]

print("SALES OF STORES 1, 4 AND 7")
print(selected_stores)

# Find the three months with the highest total sales
best_month_indices = np.argsort(monthly_sales)[-3:][::-1]

print("\nTop 3 Months:", months[best_month_indices])
print("Sales in Top 3 Months:", monthly_sales[best_month_indices])
```

```python
growth_targets = np.linspace(250000, 400000, 12)

print("MONTHLY GROWTH TARGETS")

for i in range(12):
    status = "Achieved" if monthly_sales[i] >= growth_targets[i] else "Not Achieved"

    print(
        f"Month {months[i]}: "
        f"Target = ₹{growth_targets[i]:,.2f}, "
        f"Actual = ₹{monthly_sales[i]:,}, "
        f"Status = {status}"
    )
```

```python
print("=" * 40)
print("FINAL SALES SUMMARY")
print("=" * 40)

print("Number of Stores:", len(stores))
print("Number of Months:", len(months))
print("Total Annual Sales: ₹", f"{np.sum(sales):,}")
print("Average Monthly Sales: ₹", f"{np.mean(monthly_sales):,.2f}")
print("Highest Monthly Sales: ₹", f"{np.max(monthly_sales):,}")
print("Lowest Monthly Sales: ₹", f"{np.min(monthly_sales):,}")
print("Best Store:", stores[np.argmax(store_totals)])
print("Best Month:", months[np.argmax(monthly_sales)])

print("\nProject completed successfully!")
```
